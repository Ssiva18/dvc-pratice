```python

import pyspark
print(pyspark.__version__)
import findspark
findspark.init()
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Name").getOrCreate()
print('shivaya')
```

```python
emp_data = [['Siva','Data Analyst','AXA'],['Shraddha','Pricing Analyst','AXA']]
```

```python
emp_schema = ['Names','Position','Company']
```

```python
df = spark.createDataFrame(data = emp_data,schema = emp_schema)
```

```python
df.show()
```

```python
df.rdd.getNumPartitions()
```

```python
df.where("position = 'Data Analyst'").show()
```

```python
df.printSchema()
```

```python
df.schema
```

```python
#we can select remove manipulate a column these operations are called expressions.
from pyspark.sql.functions import col,expr,cast

df.select(expr('Names'),col('company'),expr("position as pstn")).show()
```

```python
df.select('Names','Position').where("company = 'AXA'").show()
```

```python
from pyspark.sql.functions import lit
df = df.withColumn('Country',lit('India'))
```

```python
df.show()
```

```python
df = df.withColumn('Age',lit(25))
df.show()
```

```python
multiple_values = {'Place':lit('Bangalore'), ('Blood Group'): lit('B+')}
```

```python
df.withColumn(multiple_values).show()
```

```python
spark.stop()
```
